```python
y = torch.tensor([[1, 2], [3, 4]])
# 对二维张量来说，如果不设置 dim，
# 那么 torch.repeat_interleave 会先把它“展平成一维”，
# 然后再按元素重复，最后返回一维结果。
torch.repeat_interleave(y, 2)
```
